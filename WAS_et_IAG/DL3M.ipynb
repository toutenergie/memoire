# Perceptron MultiCouches (PMC)

### EXPLORATION DES DONNEES

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.datasets import cifar10

from tensorflow.keras.utils import to_categorical
```

```python
(x_train,y_train),(x_test,y_test)=cifar10.load_data()
```

```python
cifar10_classe=["avion","automabile","oiseau","chat","cerf",
                "chien","grenouille","cheval","bateau","camion"]

print("from de l'ensemble d'entrement(image):",x_train.shape)
print("from de l'ensemble d'entrement(etiquettes):",y_train.shape)
print("from de l'ensemble d'entrement(image):",x_test.shape)
print("from de l'ensemble d'entrement(etiquettes):",y_test.shape)
```

```python
# affichage de quelque image
plt.plot(figsize=(12,5))
for i range(10):
    plt.subplot(2,5,i+1)
    plt.imshow(x_train[i])
    plt.title(cifar10_classe[y_train[i][0]])
    plt.axis('off')
plt.show()
```

```python
class_distribution=[0]*10
for label in y_train:
    class_distribution[label[0]] += 1
    plt.figure(figsize=(10,4))
    plt.bar(cifar10_classe,class_distribution)
    plt.title("distribution des classes dans l'ensemble d'entrainement")
    plt.xlabel("classe")
    plt.ylabel("nonbre dimage")
    plt.xticks(rotation=45)
```

## Pretraitement des donnees

```python
x_train=x_train.astype('float32')/255.0
x_test=x_train.astype('float32')/255.0
```

## Modele PMC

```python
from tensorflow import keras
from tensorflow.keras import layers
```

```python
molede_pmc=keras.Sequential([
    # aplatissement desimage
    layers.Flatten(input_shape=(32,32,3)),
    layers.Dense(128, activation='relu'),
    layers.Dense(10, activation= 'softmax')
])

molede_pmc.compile(optimizer='adam',
                   loss='categorical_crossentropy',
                   metrics=['accuracy'])

molede_pmc.summary()
```

```python
historique=molede_pmc.fit(x_train,y_train,epochs=10,batch_size=64,
                          validation_split=0.2)
```

```python
tess_loss,test_accuracy=molede_pmc.evaluate(x_test,y_test,verbose=2)
```

```python
plt.figure(figsize=(12,4))
plt.subplot(1,2,1)
plt.plot(historique.history['loss'],label='perte d\'entraiment')
plt.plot(historique.history['val_loss'],label='perte de validation croisée')
plt.legend()
plt.title('courbe de perte')
```
